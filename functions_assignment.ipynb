```python
def find_prime_number(integer):
    for i in range(2,integer):
        if(integer%i==0):
            return False
    return True

def celsius_to_romer(celsius):
    return (celsius*(21/40))+7.5

def is_divisible(wall_size,pixel_size):
    return wall_size%pixel_size==0


def main():
    user_input=int(input("Enter an integer: "))
    if(find_prime_number(user_input)):
        print("You inputted a prime number.")
    else:
        print("You did not input a prime number.")
    temp_input=float(input("Enter a temperature in celsius: "))
    print(f" That temperature on the romer scale is {celsius_to_romer(temp_input)}.")
    wall_size=float(input("Enter wall size: "))
    pixel_size=float(input("Enter pixel size: "))
    print(is_divisible(wall_size,pixel_size))
main()





            
```
